# **🧾 Introduction to the Iris Dataset**
The Iris dataset is one of the most famous datasets in the field of **data science and machine learning**. It was introduced by the **British biologist Ronald A. Fisher** in **1936** as an example of **discriminant analysis**.

## **📦 Dataset Overview**
- Number of Samples: **150**
- Number of Features: **4**
- Number of Classes (Species): **3**

## **🏷️ Features (Columns)**
Each row represents a flower, and the columns represent:
| **Feature**     | **Description**         | **Unit** |
| ----------- | ------------------- | ---- |
| SepalLength | Length of the sepal | cm   |
| SepalWidth  | Width of the sepal  | cm   |
| PetalLength | Length of the petal | cm   |
| PetalWidth  | Width of the petal  | cm   |


## **🏷️ Target (Label)**
There are 3 species of Iris flowers in the dataset:
1. Setosa
2. Versicolor
3. Virginica

Each flower belongs to one of these species.

## **Why is it popular?**
- It's small and easy to understand.
- Ideal for classification, clustering, and visualization.
- A great starting point for testing algorithms like K-Nearest Neighbors, Logistic Regression, SVM, Decision Trees, etc.
- All features are numerical, and there's no missing data.

### 📌 6. KDE Plot of Petal Length by Species

```python
import plotly.graph_objects as go
from scipy.stats import gaussian_kde
import numpy as np

fig = go.Figure()

for species in df['species'].unique():
    data = df[df['species'] == species]['petal length (cm)']
    x = np.linspace(data.min(), data.max(), 200)
    y = gaussian_kde(data)(x)
    fig.add_trace(go.Scatter(x=x, y=y, mode='lines', name=species))

fig.update_layout(
    title='KDE Plot of Petal Length by Species',
    xaxis_title='Petal Length (cm)',
    yaxis_title='Density'
)

fig.show()
```

### **Explanation:**
This _KDE (Kernel Density Estimate)_ plot illustrates the distribution of **petal lengths** for each **Iris species**. Each curve represents a smoothed version of the petal length histogram for one species:
 - **Setosa** shows a sharp, narrow peak at a lower petal length, indicating that its petal lengths are very consistent and short.
 - **Versicolor** has a broader peak, reflecting more variation in petal lengths.
 - **Virginica** shows the highest range of petal lengths with a distribution shifted further to the right, indicating generally longer petals.

This visualization helps compare the **spread and central tendency** of petal lengths among species, and demonstrates clear **separation between species**, which is useful for classification tasks.

# 📈 **Regression Analysis (Line of Best Fit)**

### Regression Analysis: Petal Length vs Petal Width

```python
import plotly.express as px
import pandas as pd
from sklearn.datasets import load_iris
import statsmodels.api as sm
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = df['target'].map({0: 'Iris-setosa', 1: 'Iris-versicolor', 2: 'Iris-virginica'})
# Simple linear regression: Petal Width ~ Petal Length
X = df['petal length (cm)']
y = df['petal width (cm)']
X_const = sm.add_constant(X)
model = sm.OLS(y, X_const).fit()
df['predicted_width'] = model.predict(X_const)
# Scatter + Regression Line
fig = px.scatter(df, 
                 x='petal length (cm)', 
                 y='petal width (cm)', 
                 color='species',
                 opacity=0.7,
                 title='Regression Analysis: Petal Length vs Petal Width',
                 labels={
                     'petal length (cm)': 'Petal Length (cm)',
                     'petal width (cm)': 'Actual Petal Width (cm)'
                 },
                 height=600)
fig.add_scatter(x=df['petal length (cm)'], y=df['predicted_width'], 
                mode='lines', name='Regression Line', line=dict(color='black'))
fig.show()

model.summary()
```

## 📌 Interpretation of the Regression Model: Petal Length vs Petal Width

We performed a simple linear regression to examine the relationship between **Petal Length** (independent variable) and **Petal Width** (dependent variable) in the Iris dataset.

## 📊 **Pivot Table Analysis** 

### Pivot Table Analysis by Iris Species

```python
import pandas as pd
from sklearn.datasets import load_iris

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)

# Pivot Table: Mean of features per species (with observed=True to remove warning)
pivot_mean = df.pivot_table(index='species', aggfunc='mean', observed=True)
pivot_mean
```

### Multiple Summary Statistics
Instead of only using mean, include:
- **min**
- **max**
- **std** (standard deviation) — helps measure the spread/variance

```python
pivot_stats = df.pivot_table(index='species', aggfunc=['mean', 'min', 'max', 'std'], observed=True)
pivot_stats
```

### Visualize the Pivot Table Data

```python
import matplotlib.pyplot as plt

pivot_mean.plot(kind='bar', figsize=(10, 6))
plt.title('Mean Feature Values per Iris Species')
plt.ylabel('Measurement (cm)')
plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Features', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
